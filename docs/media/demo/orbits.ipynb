# Orbits

How far the earth gets from the sun over a year of steps.

```python
import math

dt, gravity = 0.001, 1.0
x, y, vx, vy = 1.0, 0.0, 0.0, 31.6
far = 0.0
for _ in range(1000):
    r = math.hypot(x, y)
    vx -= gravity * 1000 * x / r**3 * dt
    vy -= gravity * 1000 * y / r**3 * dt
    x, y = x + vx * dt, y + vy * dt
    far = max(far, math.hypot(x, y))
print(f"farthest: {far:.4f}")
```
